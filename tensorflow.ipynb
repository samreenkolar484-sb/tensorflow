```python
import tensorflow as tf
a=tf.constant([[0,1,5,6],[1,2,3,5],[2,4,6,7]])
print(a)
b=tf.transpose(a)[2]
print(b)
print(a[1:2,2:3])
print(a[1:3,2:4])
print(a[0:2,1:3])
#using negative slicing
print(a[-1:-2:-1,-3:-4:-1])
```

```python
import tensorflow as tf
a=tf.constant([[2,3,4,5],[23,45,15,8]])
print(a)
print(tf.reverse(a,axis=[1]))#first row and second row interchanged
print(tf.reverse(a,axis=[0]))# last element came to first and firts elment goes to last in every row
print(tf.reverse(a,axis=[0,1]))
print(tf.reverse(a,axis=[1,0]))
```

```python
import tensorflow as tf
a=tf.constant([[2,3,4,5],[23,45,15,8],[1,2,3,4]])
print(tf.reshape(a,[12,1]))
print(a[0],[-1,4])
print(a[0],[2,3])
```
